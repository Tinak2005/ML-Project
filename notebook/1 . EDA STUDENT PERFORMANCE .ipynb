### 1) Problem statement
The objective of this project is to analyze various factors that influence student academic performance and predict the Exam Score using Machine Learning techniques. The dataset contains academic, personal, social, and environmental factors such as Hours Studied, Attendance, Previous Scores, Motivation Level, Parental Involvement, Access to Resources, Internet Access, Tutoring Sessions, and other student-related features.

The project aims to identify patterns and relationships between these factors and student performance through Exploratory Data Analysis (EDA). Different machine learning regression models are then trained and evaluated to predict the Exam Score and identify the best-performing model.

### 3.8 Exam Score Analysis

```python
# Check Exam Score statistics

print("Minimum Exam Score:", df["Exam_Score"].min())
print("Maximum Exam Score:", df["Exam_Score"].max())
print("Average Exam Score:", df["Exam_Score"].mean())

df.head()
```

```python
# Number of students with the highest exam score

highest_score = df["Exam_Score"].max()

highest_score_count = df[
    df["Exam_Score"] == highest_score
]["Exam_Score"].count()


print(f"Highest Exam Score: {highest_score}")

print(
    f"Number of students with the highest Exam Score: "
    f"{highest_score_count}"
)
```

```python
# Number of students with a low Exam Score

low_score_threshold = 60

students_low_score = df[
    df["Exam_Score"] <= low_score_threshold
]["Exam_Score"].count()


print(
    f"Number of students with {low_score_threshold} "
    f"or less in Exam Score: {students_low_score}"
)
```

#####  Insights
 - From above values we get students have performed the worst in Maths 
 - Best performance is in reading section

### 5. Conclusions

The exploratory data analysis of the Student Performance Factors dataset helped in understanding the relationship between various academic, personal, and environmental factors and student performance. The dataset includes factors such as study hours, attendance, previous scores, motivation level, parental involvement, access to resources, and other characteristics that may influence the final Exam Score.

The visualizations showed variations in student performance across different categories and highlighted the importance of considering multiple factors rather than relying on a single variable. Numerical features such as Hours Studied, Attendance, and Previous Scores were also explored to understand their relationship with Exam Score.

Overall, the analysis indicates that student performance is influenced by a combination of academic, personal, and environmental factors. These insights provide a useful foundation for the data preprocessing and machine learning model training stages of the project.
